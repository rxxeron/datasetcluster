```python
# 2. Get Free Public URL (Choose either Native Colab or Cloudflare Tunnel - No token needed!)

# Method A: Official Colab Native Proxy URL
from google.colab.output import eval_js
colab_url = eval_js("google.colab.kernel.proxyPort(7860)")
print("\n===================================================")
print(f"🛰️ METHOD A: Official Colab Dashboard URL:")
print(f"👉 {colab_url}")
print("===================================================\n")

# Method B: Instant Public Cloudflare Tunnel (Works on phones too! Zero login!)
import subprocess, time, re
!wget -q -nc https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 -O cloudflared
!chmod +x cloudflared

# Start cloudflared in background and extract public URL
p = subprocess.Popen(["./cloudflared", "tunnel", "--url", "http://localhost:7860"], stderr=subprocess.PIPE, text=True)
public_url = None
for _ in range(15):
    line = p.stderr.readline()
    m = re.search(r'https://[a-zA-Z0-9-]+\.trycloudflare\.com', line)
    if m:
        public_url = m.group(0)
        break
    time.sleep(0.5)

if public_url:
    print("===================================================")
    print(f"🌐 METHOD B: Public Cloudflare URL (Phone & Any Browser):")
    print(f"👉 {public_url}")
    print("===================================================\n")
```
